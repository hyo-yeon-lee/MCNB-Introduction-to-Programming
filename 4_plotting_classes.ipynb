Make a bar plot with labels ['Cond1', 'Cond2', 'Cond3'], and report the mean of each condition. Show the standard deviation as errorbars.

````python
cond1 = np.random.normal(100,10,100)
cond2 = np.random.normal(120,1,100)
cond3 = np.random.normal(95,5,100)

# Replace the following line by your code
data_dict = {"cond1": cond1, "cond2": cond2, "cond3": cond3}
means = []
stds = []
labels = []
colours = ['r', 'c', 'g']

for key, values in data_dict.items():
  means.append(np.mean(values))
  stds.append(np.std(values))
  labels.append(key)

plt.plot()
plt.bar(x=labels, height = means, yerr = stds, color=colours)
plt.show()
plt.close()
````

2. Add an ```__init__``` method that initializes participant_id, age, and gender.
3. Include a ```display_info()``` method that prints the participant’s details.

````python
# REPLACE THIS LINE WITH YOUR CODE

class ParticipantData:
  def __init__(self, participant_id, age, gender):
    self.participant_id = participant_id
    self.age = age
    self.gender = gender

  def display_info(self):
    print("ID: ", self.participant_id, " | Age: ", self.age, " | Gender: ", self.gender)
````

We will now make two subclasses: one that is made for collecting experimental reaction times, and the other made for a memory task.

First, let's do the RT one.

1. Create a subclass named ```ReactionTimeExperiment``` that inherits from ```ParticipantData```.
2. Add an attribute ```self.reaction_times``` (a ```list```) to store reaction times. Initialize it empty.
3. Implement a method called ```record_reaction_time()``` to add a reaction time to the list.
4. Implement a method ```average_reaction_time()``` to calculate and return the average reaction time.

Don't forget to inherit from ```ParticipantData```. Check the lecture to see the required steps.

````python
import numpy as np

class ReactionTimeExperiment(ParticipantData):
  def __init__(self, participant_id, age, gender):
    super().__init__(participant_id, age, gender)
    self.reaction_times = []

  def record_reaction_time(self, time):
    self.reaction_times.append(time)

  def average_reaction_time(self):
    avg_rt = np.mean(self.reaction_times)
    print("Average reaction time: ", avg_rt)

````

3. Call ```average_reaction_time()``` and print the result.

````python
# REPLACE THIS LINE WITH YOUR CODE
rt1 = ReactionTimeExperiment("123", 25, "male")

record_time = np.random.randint(200, 500, 13)

for i in range(len(record_time)):
  rt1.record_reaction_time(record_time[i])

rt1.average_reaction_time()
````

Now, let's make another subclass that could be used in an experiment where participants have to remember words.

1. Create a subclass named ```MemoryTestExperiment``` that inherits from ```ParticipantData```.
2. Add an attribute called ```self.presented_words```, initialized as a ```list``` of words that the experimenter presented.
3. Add another empty attribute called ```self.recalled_words``` (a ```list```) to store recalled words. Initialize it empty.
3. Implement a method called ```record_recalled_word()``` to add a recalled word to the list. I.e., calling this method should allow you to add a word that the participant remembered to the ```self.recalled_words``` list.
4. Implement a method called ```recall_accuracy()``` to compare recalled words against presented_words and return the accuracy.

Don't forget to inherit from the ```ParticipantData``` class.

````python
# REPLACE THIS LINE WITH YOUR CODE
class MemoryTestExperiment(ParticipantData):
  def __init__(self, participant_id, age, gender, presented_words):
    super().__init__(participant_id, age, gender)
    self.presented_words = presented_words
    self.recalled_words = []

  def record_recalled_word(self, word):
    self.recalled_words.append(word)

  def recall_accuracy(self, presented_words):
    corr = set(self.presented_words) & set(self.recalled_words)
    accuracy = len(corr)/len(self.presented_words)
    return accuracy
````

````python
# Simulate reaction time experiment
participant1 = ReactionTimeExperiment(participant_id=1, age=25, gender="Female")
participant1.record_reaction_time(0.35)
participant1.record_reaction_time(0.42)
participant1.record_reaction_time(0.37)
participant1.display_info()
print(f"Average Reaction Time: {participant1.average_reaction_time()} seconds\n")

# Simulate memory test experiment
presented_words = ["apple", "banana", "grape", "orange", "pear"]
participant2 = MemoryTestExperiment(participant_id=2, age=30, gender="Male", presented_words=presented_words)
recalled_words = ["apple", "grape", "cherry"]

for word in recalled_words:
    participant2.record_recalled_word(word)

participant2.display_info()
print(f"Recall Accuracy: {participant2.recall_accuracy(presented_words) * 100}%")
````
