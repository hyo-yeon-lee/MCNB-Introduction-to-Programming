## Array creation



NumPy arrays are the basic type of object in numpy. They can be created from Python lists.

````python
my_array = np.array([1, 2, 3])
print(my_array)
````

NumPy supports array of arbitrary dimension. For example, we can create two-dimensional arrays (e.g. to store a matrix) as follows

````python
my_2d_array = np.array([[1, 2, 3], [4, 5, 6]])
print(my_2d_array)
````

We can access individual elements of a 2d-array using two indices

````python
print(my_2d_array[1, 2])
````

We can also access rows

````python
print(my_2d_array[1, :])
````

and columns

````python
print(my_2d_array[:, 2])
````

Arrays have a `shape` attribute

````python
print(my_array.shape)
print(my_2d_array.shape)
````

Contrary to Python lists, NumPy arrays must have a type and all elements of the array must have the same type.

The type determines how a number is encoded. For example, ints use all the memory to encode integers. Floats encode floating point numbers (i.e., numbers with decimals).

````python
print(my_array.dtype)
````

The main types are `int32` (32-bit integers), `int64` (64-bit integers), `float32` (32-bit real values) and `float64` (64-bit real values).

The `dtype` can be specified when creating the array

````python
my_array = np.array([1, 2, 3], dtype=np.float64)
print(my_array.dtype)
````

We can create arrays of all zeros using

````python
zero_array = np.zeros((2, 3))
print(zero_array)
````

and similarly for all ones using `ones` instead of `zeros`.

````python
one_array = np.ones((2, 3))
print(one_array)
````

We can create a range of values using

````python
print(np.arange(5))
````

or specifying the starting point

````python
print(np.arange(3, 5))
````

Another useful routine is `linspace` for creating linearly spaced values in an interval. For instance, to create 10 values in `[0, 1]`, we can use

````python
print(np.linspace(0, 1, 10))
````

Another important operation is `reshape`, for changing the shape of an array

````python
my_array = np.array([1, 2, 3, 4, 5, 6])
print(my_array.reshape(3, 2))
````

Play with these operations and make sure you understand them well.

## Basic operations

In NumPy, we express computations directly over arrays. This makes the code much more succinct.

Arithmetic operations can be performed directly over arrays. For instance, assuming two arrays have a compatible shape, we can add them as follows

````python
array_a = np.array([1, 2, 3])
array_b = np.array([4, 5, 6])
print(array_a + array_b)
````

Compare this with the equivalent computation using a for loop

````python
array_out = np.zeros_like(array_a)
for i in range(len(array_a)):
  array_out[i] = array_a[i] + array_b[i]
print(array_out)
````

Not only this code is more verbose, it will also run much more slowly.

We can also multipy each element (called element-wise multiplication).

````python
print(array_a * array_b)
````

And we can also do matrix multiplication, if thw matrix shapes are compatible (as covered in the stats class later in the semester).

For this, we use the ```@``` operator.

Note that this is different from the element-wise multiplication described above!

````python
matrixA = np.array([[1, 2, 3], [4, 5, 6]])
matrixB = np.array([[1, 2], [3, 4], [5, 6]])

print(matrixA, matrixA.shape)
print(matrixB, matrixB.shape)

mat_mul = matrixA@matrixB

print(mat_mul, mat_mul.shape)
````

NumPy contains several mathematical functions that operate on arrays in an element-wise fashion. For instance, we can do this with `np.sin()`. If we do ```np.sin(array_a))``` we will get a new array containing $[sin(a1), sin(a2), sin(a3)]$.

Let's give it a go.

````python
print(np.sin(array_a))
````

This is the case when we square numbers using the ```**``` operator, too.

````python
print(array_a**2)
````

Some other functions are applied to the array as a whole, instead of element-wise. This is the case for the vector dot product, which can be performed using `np.dot`.

>**NOTE:** This may sound a bit confusing, but in general it is quite intuitive. For example, it makes sense that the dot product, a quintessentially vectorial operation, is applied to the whole vector rather than element-wise, while the sin and square functions are applied element-wise.

````python
print(np.dot(array_a, array_b))
````

## Slicing and masking

Like Python lists, NumPy arrays support slicing

````python
print(np.arange(10)[5:])
````

We can also select only certain elements from the array

````python
x = np.arange(10)
mask = x >= 5
print(x[mask])
````
