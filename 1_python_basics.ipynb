##Variables

In python, as in any other language, ***variables*** are containers for storing data values. Variables can be created and manipulated as follows.

````python
# declare a variable
a = 10 # Note that text after a # sign is a comment, and is not considered as code

# print it
print('Initial a value') # reminder: text after # is not code, it's a comment
print(a) # comments are super important to make your code usable by future you

# overwrite it
a = 20 # and comments are also crucial to make your code usable by others

# print new content
print('Overwritten a value')
print(a)

# declare new variable b and set it to the value of a
b = a

# print b
print('Initial b value')
print(b)

# change value of a
a = 30

# notice that this changes a, but not b
print('Final a value')
print(a)
print('Final b value')
print(b)
````

## Arithmetic operations

Python supports the usual arithmetic operators: + (addition), * (multiplication), / (division), ** (power), // (integer division).

````python
a = 2
b = 4
print('a + b = ', a + b)
print('a - b = ', a * b)
print('a / b = ', a / b)
print('a // b = ', a // b)
print('a ** b = ', a ** b)
````

## Lists

Lists are a container type for ordered sequences of elements. Lists can be initialized empty

````python
my_list = []
````

or with some initial elements

````python
my_list = [1, 2, 3]
````

Lists have a dynamic size and elements can be added (appended) to them

````python
my_list.append(4)
print(my_list)
````

We can access individual elements of a list (indexing starts from 0)

````python
print(my_list[2])
````

We can access the last element of a list using -1 as an index.


````python
print(my_list[-1])
````

We can access "slices" of a list using `my_list[i:j]` where `i` is the start of the slice (again, indexing starts from 0) and `j` the end of the slice. For instance:

````python
print(my_list[1:3])
````

Omitting the first index means that the slice should run from the very beginning until the specified index.

Omitting the second index means that the slice should run until the end of the list


````python
print(my_list[:2])
print(my_list[1:])
````

We can check if an element is in the list using `in`

````python
print(5 in my_list)
````

The length of a list can be obtained using the `len` function

````python
print(len(my_list))
````

## Strings

Strings are used to store text. They can be delimited using either single quotes or double quotes

````python
string1 = "some text"
string2 = 'some other text'
````

Strings behave similarly to lists. As such we can access individual elements in exactly the same way

````python
print(string1[3])
````

and similarly for slices

````python
print(string1[5:])
````

String concatenation is performed using the `+` operator

````python
print(string1 + " " + string2)
````

## Conditionals

As their name indicates, conditionals are a way to execute code depending on whether a condition is True or False. As in other languages, Python supports `if` and `else` but `else if` is contracted into `elif`, as the example below demonstrates.

````python
my_variable = 5
if my_variable < 0:
  print("negative")
elif my_variable == 0:
  print("null")
else: # my_variable > 0
  print("positive")
````

Here `<` and `>` are the strict `less` and `greater than` operators, while `==` is the equality operator (not to be confused with `=`, the variable assignment operator). The operators `<=` and `>=` can be used for less (resp. greater) than or equal comparisons.

Contrary to other languages, blocks of code are delimited using **indentation**. That means, you need to hit tab to offset where your code is. For example, the following is incorrect:

```
if my_variable < 0:
print("negative")
elif my_variable == 0:
print("null")
else: # my_variable > 0
print("positive")
```

and the following is correct:

```
if my_variable < 0:
  print("negative")
elif my_variable == 0:
  print("null")
else: # my_variable > 0
  print("positive")```

## Loops

Loops are a way to execute a block of code multiple times. There are two main types of loops: while loops and for loops.

While loop

````python
i = 0
while i < len(my_list):
  print(my_list[i]) # note: my_list was defined in a cell earlier
  i += 1 # equivalent to i = i + 1
````

For loop

````python
for i in range(len(my_list)):
  print(my_list[i])
````

If the goal is simply to iterate over a list, we can do so directly as follows

````python
for element in my_list:
  print(element)
````
